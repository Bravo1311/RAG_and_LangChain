```python
from langchain.agents import create_agent
from langchain_core.tools import tool
from langchain_ollama import ChatOllama

@tool
def get_gnss_zones(location: str) -> str:
    """Get GNSS degradation zones for a location."""
    return f"GNSS degradation zones in {location}: urban canyons, reflective surfaces"

@tool
def analyze_image(image_path: str) -> str:
    """Analyze satellite image for signal issues."""
    return f"Analysis of {image_path}: identified multipath zones"

model = ChatOllama(model="qwen3-vl:2b", temperature=0, keep_alive="5m", num_predict=2560, num_ctx=4096)

agent = create_agent(
    model=model,
    tools=[get_gnss_zones, analyze_image],
    system_prompt="You are a GNSS signal analysis expert."
)


```

```python
result = agent.invoke({
    "messages": [{"role": "user", "content": "Analyze the urban area for GNSS multipath zones"}]
})
# print(result["messages"])
print(result["messages"][-1].content)

```
